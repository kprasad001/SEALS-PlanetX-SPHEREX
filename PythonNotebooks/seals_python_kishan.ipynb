```python
import http.client
import time
import urllib.error

import numpy as np
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.wcs import WCS
from astropy.table import Table
from astropy import units as u
from astropy.coordinates import SkyCoord
from astropy.time import Time
from astroquery.ipac.irsa import Irsa

from astroquery.ipac.irsa import Irsa

from firefly_client import FireflyClient

# The time it takes to read SPHEREx files can exceed
# astropy's default timeout limit. Increase it.
from astropy.utils.data import conf
conf.remote_timeout = 120
```

```python
ra_deg = 304.693508808
dec_deg = 42.4436872991
fov = 16 * u.arcsec

coord = SkyCoord(ra_deg, dec_deg, unit='deg')
search_radius = 1 * u.arcsec  # taking the search radius to be 1 arcsecond

#Here we prompt IRSA to return any image with coordinates that overlap with the given position and search radius
#collected from SphereX DR1 and QR3 collections
results = Irsa.query_sia(
    pos=(coord, search_radius),
    collection=['spherex_dr1', 'spherex_qr3'],
    # time=(Time('2026-01-29'), Time('2026-11-10')),   # range
)

print(len(results))

Detectors = []

for result in results:

    spectral_image_url = result['access_url']

    hdulist = fits.open(spectral_image_url, memmap=False)
    # Max number of times to retry HTTP errors.

    max_retries = 3
    for attempt in range(max_retries):
        try:
            # Load the data, reading every extension into memory so that
            # hdulist stays usable after the file is closed.
            for hdu in hdulist:
                hdu.data
            break
        except (TimeoutError, urllib.error.HTTPError, http.client.IncompleteRead):
            if attempt == max_retries - 1:
                raise
            time.sleep(10 * (attempt + 1))


    #print(result['access_url'])

    # first extension, ImageHDU
    spectral_image_header = hdulist[1].header

    detector_num = spectral_image_header.get('DETECTOR')

    if detector_num not in Detectors:
        Detectors.append(detector_num)

print("Detectors used in the data:", Detectors)
```

```python
from astropy.visualization import simple_norm
for result in range(0, len(results)):
    spectral_image_url = results['access_url'][result]
    with fits.open(spectral_image_url) as hdul:
        hdul.info()                      # list extensions
        img = hdul["IMAGE"].data
        wcs = WCS(hdul["IMAGE"].header)

    # 3. Plot
    fig = plt.figure(figsize=(7, 7))
    ax = fig.add_axes([0, 0, 1, 1], projection=wcs)  # fills the figure, no margins
    ax.imshow(img, origin="lower", cmap="gray",
            norm=simple_norm(img, "asinh", percent=99))
    ax.set_axis_off()  # hides ticks, labels, frame
    # drop the set_xlabel / set_ylabel lines

    fig.savefig("out.png", dpi=150, pad_inches=0)
    if (result == 5):
        break
plt.show()
```
